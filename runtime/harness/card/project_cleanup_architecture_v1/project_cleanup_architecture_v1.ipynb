```python
decisions = [
    {'path': 'data/', 'decision': 'keep', 'role': 'canonical data and qlib store', 'rationale': 'Active SOXL experiments need one canonical local data surface.'},
    {'path': 'soxl_backtest/', 'decision': 'keep', 'role': 'stable qlib experiment framework', 'rationale': 'Reusable framework and reference policy live here so notebooks can iterate without framework edits.'},
    {'path': 'runtime/harness/card/project_cleanup_architecture_v1/', 'decision': 'keep', 'role': 'cleanup proof', 'rationale': 'Directory-level history and deletion rationale for review.'},
    {'path': 'runtime/model_selection/card/soxlbacktest_qlib_reference_v1/', 'decision': 'keep', 'role': 'reference experiment proof', 'rationale': 'Executable SOXLBacktest parity workflow and sidecars.'},
    {'path': '.codex/skills/futu-api/', 'decision': 'keep', 'role': 'data intake only', 'rationale': 'Futu can still support data refresh/probes; not part of production execution.'},
    {'path': '.codex/skills/soxl-qlib-experiment/', 'decision': 'keep', 'role': 'single active workflow skill', 'rationale': 'Keeps future work on qlib framework plus notebook cards.'},
    {'path': 'production/', 'decision': 'delete', 'role': 'retired production execution', 'rationale': 'User chose complete removal of production surface.'},
    {'path': 'dashboard.html, dashboard_data/, web/', 'decision': 'delete', 'role': 'retired dashboard UI', 'rationale': 'Generated dashboard surface made review noisy and is out of the clean research architecture.'},
    {'path': 'src/', 'decision': 'delete', 'role': 'retired legacy harness/RL app code', 'rationale': 'Active reusable code moves under soxl_backtest; old RL/model surfaces are no longer the core architecture.'},
    {'path': 'scripts/', 'decision': 'delete', 'role': 'retired scattered entrypoints', 'rationale': 'Single-run orchestration should live in notebooks; reusable logic belongs in soxl_backtest.'},
    {'path': 'codex_exec_plans/', 'decision': 'delete', 'role': 'retired execution planning surface', 'rationale': 'Not part of the simplified qlib research loop.'},
    {'path': 'runtime/* old cards', 'decision': 'delete', 'role': 'retired historical run sprawl', 'rationale': 'Historical lessons are summarized here; old executable sprawl is removed from active review surface.'},
]
df = pd.DataFrame(decisions)
df.to_csv(CARD / 'cleanup_manifest.csv', index=False)
df
```
