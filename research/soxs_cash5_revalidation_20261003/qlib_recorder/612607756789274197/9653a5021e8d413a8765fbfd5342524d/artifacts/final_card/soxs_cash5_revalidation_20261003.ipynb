# 最终中文结论

本轮只复验此前看过的 `cash_advantage_5d` 假设，使用统一说明的复权口径、固定 5/10 特征与 StandardScaler + Ridge(alpha=10)，共 16 次拟合；没有搜索新参数或运行投资组合。数据截至 2026-09-25，先经 Qlib D 读取，再经 DatasetH.prepare 进入训练，预测与不确定性明细在 Qlib Recorder `9653a5021e8d413a8765fbfd5342524d`。

开发期 2016—2022 共 1762 行，base / 加 SOXS 的 IC 为 0.042432 / 0.053420，增量 IC 0.010987；加 SOXS 有 7/7 年 IC 为正。2023 年冻结模型在已复用历史区间共 930 行，base / 加 SOXS IC 为 0.061042 / 0.094138，增量 IC 0.033096。用户窗口按入场不早于 2026-08-03、H5 退出不晚于 2026-09-25 开盘计算，共 34 行，base / 加 SOXS 的 IC 分别为 -0.13277 / -0.08235，增量 IC 0.050420；两者在该小窗口仍为负，这是 H5 样本，不能沿用 H1 的行数。年间差异见 `point_metrics.csv`，本轮没有另做 regime 搜索。

预注册的信息 / 增量 / 经济门均失败；总判定 `end_route_no_further_search`。旧 5/7 正年与 p05 规则兼容诊断通过，但开发期 20 日块的完整特征 IC 95% 区间为 [-0.01428, 0.13193]，信息门仍失败。两窗口在 20/60 日块下的配对增量 IC 与 MSE 改善区间全部跨零，不能确认加入 SOXS 带来稳定增量。双窗口、20/60 日块、各 2,000 次同索引配对自助结果见 `bootstrap_summary.csv`；任一未定义抽样导致该区间未定义并失败。10/25bp 成本诊断采用卖出系数 1-c、买入资金系数 1/(1+c)，仅评价相互重叠的独立 SOXX→现金→SOXX 事件，不复利、不年化、不视作投组收益。25bp 单边费用下，加 SOXS 模型在开发期触发 27 个事件，平均净现金优势 +37.95bp；复用期触发 57 个事件，均值变为 -53.29bp，两期自助区间均跨零。10bp 下的相应均值也分别为 -1.97bp 和 -25.78bp。现有证据未证明稳定、可交易的净优势。

不晋升：全部历史均已观察，选择该假设本身受旧阳性结果影响，不能当作新样本外证据。2023 年模型修复了旧外层训练的年末标签穿越，使用同样 20 行剔除及 H5 成熟日期严格早于年界；因此不声称精确重现旧外层数值。三序列是回溯至 2026-09-25 的复权版本，未来分红的加法 B 可改变历史比率；训练切片和标签门不能证明完整的历史时点复权可得性。SOXS 早期 709 行依赖第二来源，并补充 Futu 缺失的早期事件；详见 `source_audit.json`，不等同于 SDK QFQ 或生产级权威数据。
