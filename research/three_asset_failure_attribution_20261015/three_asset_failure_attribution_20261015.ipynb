## 胜负区间与归因

年度按trade_date而非旧卡signal_date分组；2023左截断、2026右截断，只有2024/2025完整。因此不根据四个标签直接宣称年度门通过。小配置头2bp四年相对对数收益依次-0.12262、+0.06078、+0.00086、-0.05261；每天执行为-0.11925、+0.10962、+0.00124、-0.05160。每天执行改善集中于2024，2023和2026主要缺口依然存在。季度全部保留在quarterly_attribution.csv。

小配置头2bp相对SOXX对数财富-0.11359，由15%参考现金-0.21067、额外现金-0.26077、SOXL相对贡献+0.37559、恢复0、费用-0.02015及基准/原生微小残差组成。每天执行对应-0.05999，其中现金两项-0.43910，SOXL+0.39796。15%仅是归因分界，并非本任务新的现金硬约束。对数贡献不是收益百分点，也不是去除现金后能实现的收益。

使用源signal_date已知的SOXX相对20日均线正负作固定描述分层：小配置头2bp正趋势586日相对对数贡献+0.12667，非正趋势358日-0.24026；每天执行分别+0.12646/-0.18646。0/2/5bp都保留这个方向。正趋势中SOXL贡献约+0.372，非正趋势接近零，现金损失抵消不足。但分层已在复用历史上观察，不能据此直接做趋势开关或宣称新增预测信息。

风险状态是各自账户产生的内生标签；转换日持仓可能来自上一阶段，不视为可随机比较的市场组。09:46开盘决策对应09:47成交；recovery_action是10:32后的已实现动作，不是开盘可见状态。本批恢复动作全部为零。状态汇总只有非连续会计贡献，不计算虚构连续状态组合的回撤。
