```python
import sys
from pathlib import Path
import qlib
from qlib.data import D
from qlib.contrib.evaluate import risk_analysis
from qlib.workflow import R

ROOT = Path.cwd()
while not (ROOT / 'src').exists():
    if ROOT == ROOT.parent:
        raise RuntimeError('could not locate repository root')
    ROOT = ROOT.parent
sys.path.insert(0, ROOT.as_posix())
from src.workflow.soxx_hidden_dispersion import run_soxx_hidden_dispersion_card

qlib.init(provider_uri=(ROOT / 'data' / 'qlib' / 'soxl').as_posix(), region='us', kernels=1)
_ = D.features(['SOXL', 'SOXX'], ['$open', '$vix'], start_time='2019-01-02', freq='day')
assert callable(R.start)
assert callable(risk_analysis)
summary = run_soxx_hidden_dispersion_card(root=ROOT)
assert Path(R.get_uri()).resolve() == (ROOT / summary['recorder_uri']).resolve()
summary
```
