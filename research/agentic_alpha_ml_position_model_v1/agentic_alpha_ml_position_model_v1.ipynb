# Agentic Alpha ML Position Model v1

Hypothesis: replace the unstable constrained-weight position rule with supervised ML position signals. Model and position thresholds are selected only on 2025 validation data; 2026 is a holdout report window.

```python

from __future__ import annotations

import json
import shutil
import sys
import warnings
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd
import qlib
from qlib.data import D
from qlib.workflow import R

warnings.filterwarnings('ignore', message='X does not have valid feature names')

ROOT_CANDIDATES = [Path.cwd(), Path('/Users/rziiiii/dev/workspace')]
ROOT = next(
    path
    for base in ROOT_CANDIDATES
    for path in [base, *base.parents]
    if (path / 'AGENTS.md').exists() and (path / 'SPEC.md').exists()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.backtest import evaluate_ml_position_candidates
from src.factor import (
    build_canary_sprint_frame,
    build_canary_trend_component,
    build_session_effect_frame,
    build_signed_overnight_component,
    build_tsmom_stop_component,
)
from soxl_backtest.reference import replay_soxlbacktest_reference

EXPERIMENT = 'agentic_alpha_ml_position_model_v1'
CARD = ROOT / 'research' / EXPERIMENT
PROVIDER_URI = (ROOT / 'data' / 'qlib' / 'soxl').resolve()
RECORDER_URI = (CARD / 'qlib_recorder').resolve()
SOURCE_START = '2010-03-11'
DATA_START = '2019-01-02'
DATA_END = '2026-06-09'
SOURCE_END = '2026-06-23'
TRAIN_WINDOW = ('2019-01-02', '2024-12-31')
VALIDATION_WINDOW = ('2025-01-02', '2025-12-31')
TEST_WINDOW = ('2026-01-02', '2026-06-09')
MAX_DRAWDOWN_FLOOR = -20.0
MAX_POSITION = 1.0
MIN_VALIDATION_TRADE_COUNT = 10
MIN_VALIDATION_AVG_ABS_POSITION = 0.05
COST_PER_TURNOVER = 0.0003

CARD.mkdir(parents=True, exist_ok=True)
if RECORDER_URI.exists():
    shutil.rmtree(RECORDER_URI)
RECORDER_URI.mkdir(parents=True, exist_ok=True)

workflow_yaml = f'''experiment: {EXPERIMENT}
card_type: ml_position_model_search
qlib:
  provider_uri: {PROVIDER_URI.as_posix()}
  recorder_uri: {RECORDER_URI.as_posix()}
  market: all
  region: us
segments:
  train:
    start: {TRAIN_WINDOW[0]}
    end: {TRAIN_WINDOW[1]}
  validation:
    start: {VALIDATION_WINDOW[0]}
    end: {VALIDATION_WINDOW[1]}
  test:
    start: {TEST_WINDOW[0]}
    end: {TEST_WINDOW[1]}
ml_position_model:
  selection_segment: validation
  report_segment: test
  max_position: {MAX_POSITION}
  min_validation_trade_count: {MIN_VALIDATION_TRADE_COUNT}
  min_validation_avg_abs_position: {MIN_VALIDATION_AVG_ABS_POSITION}
  validation_max_drawdown_floor: {MAX_DRAWDOWN_FLOOR}
artifact_policy:
  policy: slim_card_recorder_first
  machine_entrypoint: summary.json
'''
(CARD / 'workflow.yaml').write_text(workflow_yaml, encoding='utf-8')
manifest = json.loads((PROVIDER_URI / 'qlib_store_manifest.json').read_text(encoding='utf-8'))
print(json.dumps({'provider_as_of': manifest['as_of'], 'experiment': EXPERIMENT}, indent=2, sort_keys=True))
```

```python

def build_ml_feature_frame() -> tuple[pd.DataFrame, list[str], list[str]]:
    qlib.init(provider_uri=PROVIDER_URI.as_posix(), region='us', kernels=1, clear_mem_cache=True)
    daily_qlib_frame = D.features(
        ['SOXL', 'SOXX'],
        ['$open', '$high', '$low', '$close', '$volume', '$vix'],
        start_time=SOURCE_START,
        end_time=SOURCE_END,
        freq='day',
    )
    session_features = build_session_effect_frame(daily_qlib_frame, start=DATA_START, end=DATA_END, min_history=20)
    canary_features = build_canary_sprint_frame(daily_qlib_frame, start=DATA_START, end=DATA_END)
    components = [
        build_tsmom_stop_component(
            daily_qlib_frame,
            component='tsmom_soxx_ret10_th0_lowhv_q70_s1.0_stop0.02',
            start=DATA_START,
            end=DATA_END,
            score_name='soxx_ret10',
            threshold=0.0,
            regime_filter='lowhv',
            scale=1.0,
            stop_loss=0.02,
            lowhv_quantile=0.70,
        ),
        build_tsmom_stop_component(
            daily_qlib_frame,
            component='tsmom_soxx_ret252_th0.15_vix30_s0.75_stop0.03',
            start=DATA_START,
            end=DATA_END,
            score_name='soxx_ret252',
            threshold=0.15,
            regime_filter='vix30',
            scale=0.75,
            stop_loss=0.03,
            lowhv_quantile=0.70,
        ),
        build_tsmom_stop_component(
            daily_qlib_frame,
            component='tsmom_soxx_ret10_th0_lowhv_q70_s0.75_stop0.02',
            start=DATA_START,
            end=DATA_END,
            score_name='soxx_ret10',
            threshold=0.0,
            regime_filter='lowhv',
            scale=0.75,
            stop_loss=0.02,
            lowhv_quantile=0.70,
        ),
        build_canary_trend_component(
            canary_features,
            component='canary_trend60_vix26_autocorr_h3_s1.0',
            start=DATA_START,
            end=DATA_END,
            trend_threshold=0.06,
            vix_max=26.0,
            hold_days=3,
            require_positive_autocorr=True,
            scale=1.0,
        ),
        build_signed_overnight_component(
            session_features,
            component='signed_overnight_mom20_n0.002_soxl60_vix45_s0.5',
            start=DATA_START,
            end=DATA_END,
            neutral_band=0.002,
            scale=0.5,
            momentum_window=20,
            trend_filter='soxl60',
            vix_max=45.0,
        ),
    ]
    component_frame = pd.concat(components, ignore_index=True)
    component_exposure = (
        component_frame.pivot_table(index='datetime', columns='component', values='exposure', aggfunc='sum')
        .sort_index()
        .fillna(0.0)
    )

    normalized = daily_qlib_frame.reset_index()
    normalized['instrument'] = normalized['instrument'].astype(str).str.upper()
    normalized['datetime'] = pd.to_datetime(normalized['datetime']).dt.normalize()
    soxl = normalized.loc[
        normalized['instrument'] == 'SOXL',
        ['datetime', '$open', '$high', '$low', '$close', '$volume', '$vix'],
    ].sort_values('datetime')
    soxx = normalized.loc[
        normalized['instrument'] == 'SOXX', ['datetime', '$close', '$volume']
    ].sort_values('datetime').rename(columns={'$close': 'soxx_close', '$volume': 'soxx_volume'})
    panel = soxl.merge(soxx, on='datetime', how='inner').rename(
        columns={
            '$open': 'soxl_open',
            '$high': 'soxl_high',
            '$low': 'soxl_low',
            '$close': 'soxl_close',
            '$volume': 'soxl_volume',
            '$vix': 'vix',
        }
    ).sort_values('datetime').reset_index(drop=True)
    panel['forward_return'] = panel['soxl_close'].pct_change().shift(-1)
    for window in [1, 2, 3, 5, 10, 20, 40, 60, 120, 126, 200, 252]:
        panel[f'soxl_ret{window}'] = panel['soxl_close'].pct_change(window)
        panel[f'soxx_ret{window}'] = panel['soxx_close'].pct_change(window)
    for window in [10, 20, 40, 60, 120, 252]:
        panel[f'soxl_hv{window}'] = panel['soxl_close'].pct_change().rolling(
            window, min_periods=min(window, 20)
        ).std() * np.sqrt(252)
    for window in [20, 60, 120, 252]:
        panel[f'soxl_dd{window}'] = panel['soxl_close'] / panel['soxl_close'].rolling(
            window, min_periods=min(window, 20)
        ).max() - 1.0
    panel['vix_chg1'] = panel['vix'].pct_change()
    panel['vix_chg5'] = panel['vix'].pct_change(5)
    panel['volume_ratio20'] = panel['soxl_volume'] / panel['soxl_volume'].rolling(20, min_periods=10).mean() - 1.0

    frame = component_exposure.reset_index().merge(panel, on='datetime', how='inner')
    component_cols = list(component_exposure.columns)
    old_selected_model = json.loads((ROOT / 'research' / 'agentic_alpha_ensemble_search_v1' / 'selected_model.json').read_text(encoding='utf-8'))
    old_weights = {str(key): float(value) for key, value in old_selected_model['weights'].items()}
    frame['old_raw_target'] = frame[component_cols].mul(
        pd.Series({column: old_weights.get(column, 0.0) for column in component_cols}),
        axis=1,
    ).sum(axis=1)
    frame['old_raw_cap1'] = frame['old_raw_target'].clip(0.0, 1.0)
    frame['canary_base'] = frame['canary_trend60_vix26_autocorr_h3_s1.0'].clip(0.0, 1.0)
    feature_cols = component_cols + [
        'old_raw_target',
        'old_raw_cap1',
        'canary_base',
        'soxl_ret1',
        'soxl_ret2',
        'soxl_ret5',
        'soxl_ret10',
        'soxl_ret20',
        'soxl_ret60',
        'soxx_ret10',
        'soxx_ret20',
        'soxx_ret60',
        'soxx_ret120',
        'soxx_ret252',
        'soxl_hv20',
        'soxl_hv60',
        'soxl_dd20',
        'soxl_dd60',
        'soxl_dd120',
        'vix',
        'vix_chg1',
        'vix_chg5',
        'volume_ratio20',
    ]
    frame = frame.loc[(frame['datetime'] >= pd.Timestamp(DATA_START)) & (frame['datetime'] <= pd.Timestamp(DATA_END))]
    frame = frame.replace([np.inf, -np.inf], np.nan).dropna(subset=['forward_return']).reset_index(drop=True)
    return frame, feature_cols, component_cols

ml_frame, feature_cols, component_cols = build_ml_feature_frame()
print(json.dumps({'rows': int(len(ml_frame)), 'features': int(len(feature_cols)), 'component_features': component_cols}, indent=2, sort_keys=True))
```

```python

position_specs: list[dict[str, Any]] = []
for mode in ['probability', 'binary']:
    for threshold in [0.50, 0.55, 0.60, 0.65]:
        for scale in [0.5, 0.75, 1.0]:
            for rebalance in [0.10, 0.20, 0.25]:
                position_specs.append(
                    {
                        'position_mode': mode,
                        'threshold': threshold,
                        'scale': scale,
                        'rebalance_threshold': rebalance,
                        'min_position': 0.0,
                        'max_position': MAX_POSITION,
                    }
                )
for base_position_col in ['old_raw_cap1', 'canary_base']:
    for mode in ['base_gate', 'base_probability']:
        for threshold in [0.50, 0.55, 0.60, 0.65]:
            for rebalance in [0.10, 0.20, 0.25]:
                position_specs.append(
                    {
                        'position_mode': mode,
                        'threshold': threshold,
                        'scale': 1.0,
                        'base_position_col': base_position_col,
                        'rebalance_threshold': rebalance,
                        'min_position': 0.0,
                        'max_position': MAX_POSITION,
                    }
                )

model_specs = [
    {'model': 'logistic_regression'},
    {'model': 'hist_gradient_boosting_classifier'},
    {'model': 'lgbm_classifier'},
    {'model': 'ridge'},
    {'model': 'hist_gradient_boosting_regressor'},
    {'model': 'lgbm_regressor'},
]

candidate_metrics, daily_positions = evaluate_ml_position_candidates(
    ml_frame,
    feature_cols=feature_cols,
    model_specs=model_specs,
    train_window=TRAIN_WINDOW,
    validation_window=VALIDATION_WINDOW,
    test_window=TEST_WINDOW,
    position_specs=position_specs,
    cost_per_turnover=COST_PER_TURNOVER,
    min_validation_trade_count=MIN_VALIDATION_TRADE_COUNT,
    min_validation_avg_abs_position=MIN_VALIDATION_AVG_ABS_POSITION,
    validation_max_drawdown_floor=MAX_DRAWDOWN_FLOOR,
)
selected_validation_row = candidate_metrics.loc[
    (candidate_metrics['selected']) & (candidate_metrics['segment'] == 'validation')
].iloc[0].to_dict()
selected_test_row = candidate_metrics.loc[
    (candidate_metrics['selected']) & (candidate_metrics['segment'] == 'test')
].iloc[0].to_dict()
oracle_test_row = candidate_metrics.loc[candidate_metrics['segment'] == 'test'].sort_values(
    'selection_objective_score', ascending=False
).iloc[0].to_dict()

legacy_replay = replay_soxlbacktest_reference(start=TEST_WINDOW[0], end=TEST_WINDOW[1])
legacy_metrics = legacy_replay.metrics
constrained_summary = json.loads((ROOT / 'research' / 'agentic_alpha_constrained_weight_search_v1' / 'summary.json').read_text(encoding='utf-8'))
constrained_reference = constrained_summary['selected_candidate']
raw_reference = constrained_summary['raw_position_reference']

print(candidate_metrics.loc[candidate_metrics['segment'] == 'validation'].sort_values('selection_objective_score', ascending=False).head(8)[[
    'candidate', 'return_pct', 'max_drawdown_pct', 'avg_abs_position', 'trade_count', 'next_day_reversal_count', 'total_turnover', 'selection_objective_score', 'selection_eligible'
]].to_string(index=False))
print(candidate_metrics.loc[candidate_metrics['segment'] == 'test'].sort_values('selection_objective_score', ascending=False).head(8)[[
    'candidate', 'return_pct', 'max_drawdown_pct', 'avg_abs_position', 'trade_count', 'next_day_reversal_count', 'total_turnover', 'selection_objective_score'
]].to_string(index=False))
```

```python

def jsonable(value: Any) -> Any:
    if isinstance(value, dict):
        return {str(key): jsonable(item) for key, item in value.items()}
    if isinstance(value, list):
        return [jsonable(item) for item in value]
    if isinstance(value, tuple):
        return [jsonable(item) for item in value]
    if isinstance(value, (np.bool_, bool)):
        return bool(value)
    if isinstance(value, (np.integer, int)):
        return int(value)
    if isinstance(value, (np.floating, float)):
        numeric = float(value)
        return None if not np.isfinite(numeric) else numeric
    if pd.isna(value):
        return None
    return value


def compact_metric_row(row: dict[str, Any]) -> dict[str, Any]:
    keys = [
        'candidate',
        'model',
        'model_task',
        'position_mode',
        'threshold',
        'scale',
        'return_pct',
        'max_drawdown_pct',
        'return_to_drawdown',
        'selection_objective_score',
        'avg_abs_position',
        'max_position',
        'min_position',
        'days_above_100',
        'days_below_0',
        'total_turnover',
        'trade_count',
        'next_day_reversal_count',
        'avg_holding_days',
        'selection_eligible',
        'selection_gate',
        'params',
    ]
    return {key: jsonable(row.get(key)) for key in keys if key in row}

candidate_metrics_path = CARD / 'ml_position_candidate_metrics.csv'
selected_model_path = CARD / 'ml_position_selected_model.json'
ml_summary_path = CARD / 'ml_position_summary.json'
benchmark_path = CARD / 'backtest_benchmark_comparison.csv'
failure_modes_path = CARD / 'failure_modes.md'
daily_positions_tmp = CARD / '.ml_position_daily_positions.csv.gz'

candidate_metrics.to_csv(candidate_metrics_path, index=False)
daily_positions.to_csv(daily_positions_tmp, index=False, compression='gzip')
selected_model_payload = {
    'candidate': selected_test_row['candidate'],
    'selected_by': 'validation_selection_objective_score',
    'validation': compact_metric_row(selected_validation_row),
    'test': compact_metric_row(selected_test_row),
}
selected_model_path.write_text(json.dumps(selected_model_payload, indent=2, sort_keys=True), encoding='utf-8')

benchmark = pd.DataFrame(
    [
        {
            'benchmark': 'selected_ml_validation_model',
            'return_pct': float(selected_test_row['return_pct']),
            'max_drawdown_pct': float(selected_test_row['max_drawdown_pct']),
            'max_position': float(selected_test_row['max_position']),
            'avg_abs_position': float(selected_test_row['avg_abs_position']),
            'trade_count': int(selected_test_row['trade_count']),
            'next_day_reversal_count': int(selected_test_row['next_day_reversal_count']),
        },
        {
            'benchmark': 'oracle_test_best_ml_candidate',
            'return_pct': float(oracle_test_row['return_pct']),
            'max_drawdown_pct': float(oracle_test_row['max_drawdown_pct']),
            'max_position': float(oracle_test_row['max_position']),
            'avg_abs_position': float(oracle_test_row['avg_abs_position']),
            'trade_count': int(oracle_test_row['trade_count']),
            'next_day_reversal_count': int(oracle_test_row['next_day_reversal_count']),
        },
        {
            'benchmark': 'constrained_weight_reference',
            'return_pct': float(constrained_reference['return_pct']),
            'max_drawdown_pct': float(constrained_reference['max_drawdown_pct']),
            'max_position': float(constrained_reference['max_position']),
            'avg_abs_position': None,
            'trade_count': int(constrained_reference['trade_count']),
            'next_day_reversal_count': int(constrained_reference['next_day_reversal_count']),
        },
        {
            'benchmark': 'raw_target_position_reference',
            'return_pct': float(raw_reference['return_pct']),
            'max_drawdown_pct': float(raw_reference['max_drawdown_pct']),
            'max_position': float(raw_reference['max_position']),
            'avg_abs_position': float(raw_reference['avg_abs_position']),
            'trade_count': int(raw_reference['trade_count']),
            'next_day_reversal_count': int(raw_reference['next_day_reversal_count']),
        },
        {
            'benchmark': 'soxl_backtest_reference',
            'return_pct': float(legacy_metrics['rulebase_return_pct']),
            'max_drawdown_pct': -float(legacy_metrics['rulebase_max_drawdown_pct']),
            'max_position': float(legacy_replay.daily['position'].max()),
            'avg_abs_position': float(legacy_replay.daily['position'].abs().mean()),
            'trade_count': int(legacy_metrics['trade_count']),
            'next_day_reversal_count': None,
        },
        {
            'benchmark': 'soxl_buyhold',
            'return_pct': float(legacy_metrics['soxl_buyhold_return_pct']),
            'max_drawdown_pct': None,
            'max_position': 1.0,
            'avg_abs_position': 1.0,
            'trade_count': 1,
            'next_day_reversal_count': None,
        },
    ]
)
benchmark.to_csv(benchmark_path, index=False)
model_families = [
    'HistGradientBoostingClassifier',
    'HistGradientBoostingRegressor',
    'LGBMClassifier',
    'LGBMRegressor',
    'LogisticRegression',
    'Ridge',
]
ml_position_summary = {
    'workflow': EXPERIMENT,
    'goal_achieved': False,
    'candidate_count': int(candidate_metrics['candidate'].nunique()),
    'metric_rows': int(len(candidate_metrics)),
    'feature_count': int(len(feature_cols)),
    'feature_cols': feature_cols,
    'component_cols': component_cols,
    'model_families': model_families,
    'selection_gate': {
        'selection_segment': 'validation',
        'report_segment': 'test',
        'min_validation_trade_count': MIN_VALIDATION_TRADE_COUNT,
        'min_validation_avg_abs_position': MIN_VALIDATION_AVG_ABS_POSITION,
        'validation_max_drawdown_floor': MAX_DRAWDOWN_FLOOR,
        'max_position': MAX_POSITION,
    },
    'selected_candidate': selected_model_payload,
    'oracle_test_best_candidate': compact_metric_row(oracle_test_row),
    'constrained_weight_reference': constrained_reference,
    'raw_position_reference': raw_reference,
    'legacy_soxlbacktest_benchmark': legacy_metrics,
    'underperformance': {
        'selected_vs_soxlbacktest_return_gap_pct': float(selected_test_row['return_pct'] - legacy_metrics['rulebase_return_pct']),
        'selected_vs_constrained_return_gap_pct': float(selected_test_row['return_pct'] - constrained_reference['return_pct']),
        'oracle_test_vs_soxlbacktest_return_gap_pct': float(oracle_test_row['return_pct'] - legacy_metrics['rulebase_return_pct']),
    },
}
ml_summary_path.write_text(json.dumps(jsonable(ml_position_summary), indent=2, sort_keys=True), encoding='utf-8')
failure_modes_path.write_text(
    '# Failure Modes\n\n'
    '- validation-selected ML candidate is stable but too under-invested in 2026, so it misses the SOXL trend leg.\n'
    '- oracle test-top ML candidate is diagnostic only because it uses 2026 ranking; it still fails to beat soxl_backtest.\n'
    '- Features are built from previously researched agentic components, so this is not independent proof of a production architecture.\n'
    '- Single-asset daily ML sample is small; prefer walk-forward paper simulation before any promotion.\n',
    encoding='utf-8',
)

R.set_uri(RECORDER_URI.as_posix())
with R.start(experiment_name=EXPERIMENT):
    recorder = R.get_recorder()
    recorder.log_params(
        provider_uri=PROVIDER_URI.as_posix(),
        train_start=TRAIN_WINDOW[0],
        train_end=TRAIN_WINDOW[1],
        validation_start=VALIDATION_WINDOW[0],
        validation_end=VALIDATION_WINDOW[1],
        test_start=TEST_WINDOW[0],
        test_end=TEST_WINDOW[1],
        feature_count=int(len(feature_cols)),
        candidate_count=int(candidate_metrics['candidate'].nunique()),
        min_validation_trade_count=MIN_VALIDATION_TRADE_COUNT,
        min_validation_avg_abs_position=MIN_VALIDATION_AVG_ABS_POSITION,
        validation_max_drawdown_floor=MAX_DRAWDOWN_FLOOR,
        cost_per_turnover=COST_PER_TURNOVER,
    )
    recorder.log_metrics(
        selected_test_return_pct=float(selected_test_row['return_pct']),
        selected_test_max_drawdown_pct=float(selected_test_row['max_drawdown_pct']),
        oracle_test_best_return_pct=float(oracle_test_row['return_pct']),
        constrained_weight_return_pct=float(constrained_reference['return_pct']),
        soxlbacktest_return_pct=float(legacy_metrics['rulebase_return_pct']),
    )
    for path in [candidate_metrics_path, selected_model_path, ml_summary_path, benchmark_path, failure_modes_path]:
        recorder.log_artifact(path.as_posix(), artifact_path='agentic_alpha_ml_position_model')
    recorder.log_artifact(daily_positions_tmp.as_posix(), artifact_path='agentic_alpha_ml_position_model')
    recorder_id = recorder.id
    recorder_local_dir = Path(recorder.uri).resolve()

daily_positions_tmp.unlink(missing_ok=True)
ml_position_summary['recorder_id'] = recorder_id
ml_position_summary['recorder_local_dir'] = recorder_local_dir.as_posix()
ml_position_summary['recorder_artifacts'] = sorted(
    path.relative_to(recorder_local_dir).as_posix()
    for path in recorder_local_dir.rglob('*')
    if path.is_file()
)
ml_summary_path.write_text(json.dumps(jsonable(ml_position_summary), indent=2, sort_keys=True), encoding='utf-8')
print(json.dumps({'recorder_id': recorder_id, 'selected_test_return_pct': float(selected_test_row['return_pct']), 'oracle_test_best_return_pct': float(oracle_test_row['return_pct'])}, indent=2, sort_keys=True))
```
