# 候选TTM来源覆盖结论

使用完整47CIK事实、Qlib日历和按公开日期可用的NPORT篮子，执行货币利润的年度直读或“去年全年＋今年YTD－去年同长度YTD”。每项保留组成accession、财务区间、申报日、下一交易日可用日、币种和会计标签，最新披露旧报告不会倒退篮子。后续修订只能改变其公开之后的计算。

实际覆盖90个月末（2019-01至2026-06），其中80个月已有公开持仓；保留前10个月缺持仓。生成3210条证券/截止日明细。候选收入权重77.84%–99.83%，只有['2025-03-31', '2025-04-30']达到95%；28个持仓披露截止日均未达95%。这不是可交易EP，也不是策略测试。年度直读和派生TTM分开保留，负利润不删除，未覆盖权重不归一。

后续排查发现具体实现缺口：AVGO的季度事实采用us-gaap:ProfitLoss，而首版冻结标签只包含us-gaap:NetIncomeLoss、ifrs-full:ProfitLoss，所以不能把本结果说成数据不存在或整体估值路线失败。TSM/UMC的该来源则主要提供年度美元便利换算值，季度覆盖确有不足，需要原始季度资料和币种/ADR口径。这些限制须分别处理，不靠降低95%或36个月门槛掩盖。

会计口径仍是候选：年度和YTD的最新可用修订可能存在会计基础差异，NetIncomeLoss与IFRSProfitLoss不自动等于归属普通股股东利润。尚未完成股份、ADR、汇率、拆股和价格口径验证，合格EP行数0、没有价格收益、标签、训练或回测、不晋升。下一步应独立登记AVGO实际标签扩展并补齐外国发行人季度来源，再评估真正的月度EP历史是否可用；整体稳定超额目标没有达成或被本卡关闭。
