## 界的含义与峰谷

固定长仓在每分钟内满足L=cash+Σ(q×asset low)、U=cash+Σ(q×asset high)。运行U高水位与当前L组合给出保守回撤上界，固定费用跳变及成交前后精确NAV也按时间顺序进入高水位。两资产high/low可能不同步，同一分钟的高低先后也未知；加权资产极值不是可实现的同时组合价格，上界不是实际实现的组合OHLC或收益路径。

候选三成本的已观察最高NAV标记位于2026-06-22 10:15分钟close，最深回撤位于2026-07-29 16:00 close。保守界的峰值使用06-22 10:19–10:20分钟包络，谷值使用07-29 15:59–16:00包络；这些区间不指定未知的极值精确发生时刻。各成本及SOXX的全部最坏峰谷记录见worst_episodes.csv。

若观察回撤≤30%但上界>30%，只能判定分钟内次序未解决，不会据上界制造实际违约。本批候选没有这种中间情形。该界只对原价格代理与固定账本的RTH范围成立，不升级为真实券商成交、真实股息现金流、历史首收或全天风险认证。
