## Takeaways

### 中文结论

- 做了什么：检验价格路径耗竭、杠杆脆弱性、固定五龙头结构代理和 SOXL/SOXX 盘中路径，共 21 个透明因子，不使用预测模型。
- 关键指标：2023–2026 SOXL 高位样本 461 条、33 次 breakdown；price-only AUC 0.705、AP lift 1.75x、Q5-Q1 13.1%。
- regime/window：2023 和 2024 是可用的年度事件窗口；2025/2026 只有 3/2 次事件，年度结果只作诊断。
- 晋升决定：`not_retained_no_incremental_factor_evidence`。开发期唯一入选的 15 分钟 range-ratio 因子在外层反转为 AUC 0.401、Q5-Q1 -6.4%、price-only 条件增量 -3.9%。其余三族没有因子通过开发门槛。
- 风险：固定五龙头存在生存偏差；外部龙头 15 分钟历史不足；单因子外层排名属于诊断，不能反向用于重选。
- 下一步：保留 price-only 作为风险状态基线；若继续优化，应改标签为近端路径失速/回撤深度并做多尺度事件定义，而不是继续给这些失败因子调权重。
