本支线完成冻结的20个公共来源动作：4次检索、16份文档请求，全部原始响应和hash保留；没有生成价格序列、收益标签、模型或账户。该支线不阻塞年度估值主线。

两个实际Nasdaq历史导出请求（2024-09-20、2025-09-19）均返回有效Excel文件，各30家公司。但真实字段只有Company Name和Security Symbol，文件标题虽为Weightings，正文并无权重，也无独立的历史asof标记。保留为“带请求日期的历史成分响应”，不能升级成完整调仓权重。两份名单共有26个名字；4个旧名单独有、4个新名单独有是跨年差异，不能归因于单次调仓。

取得了一份真实历史的部分权重：Nasdaq于2025年10月发布的研究文档，在第4页给出2025-09-30的前10名、0.1个百分点精度的权重图及其余39.1%合计。它是调仓后的月末快照，缺其余20名的个别权重和精确发表日，不是调仓前公告。另一个事实陷阱是固定FS_SOX.pdf URL：搜索索引显示2025年日期，而实际下载及视觉核对的正文已是2026-09-30；不能把当前文件回填历史。

Nasdaq权重页和月末历史页明确提示完整访问需要登录/订阅，免费页面与历史Excel只提供成分名称。ICE官方公告/调仓页面转至ICE Index Platform，本次未取得两个目标调仓的完整权重文件或公告时点。该证据只说明已检查入口的访问边界，不证明所有其他公开档案绝对不存在。

复用的SOXX SEC基金持仓分别为2024-09-30和2025-09-30各30个股票，最早披露均在当年11月26日；这些是有金额的事后基金披露，不能当成9月调仓前指数权重。没有用等权或当前成分补齐未知权重。

时间/身份分开登记：2021年SOXX从PHLX转向ICE的公告与预计生效日；2023年10月13日公告、11月3日生效的ICE→NYSE名称更改（符号不变）；SOX于2024-04-22生效的12/10/8及4%上限方法变化。上限不是实际权重，名称变化也不证明方法全历史不变。未来若比较总收益，SOX价格版本与XSOX总收益版本须区分，本卡没有计算收益。

结论：合格完整调仓权重集合0、合格目标公告时点0，不晋升。最小下一输入是两个冻结目标日期的全体权重，以及公告、实际生效、资料公开时点；需要已有授权历史数据或其他独立带时点的官方文件。保留当前可审计的成分响应、部分历史图和基金披露，按20次预算结束本来源支线，稳定超额研究与年度主线继续。
