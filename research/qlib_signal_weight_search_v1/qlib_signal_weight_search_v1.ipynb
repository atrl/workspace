```python
from pathlib import Path
import json
import sys

import qlib
from qlib.data import D
from qlib.workflow import R
from qlib.data.dataset import DatasetH
from qlib.data.dataset.handler import DataHandlerLP
from qlib.workflow.record_temp import SignalRecord, SigAnaRecord, PortAnaRecord
from qlib.contrib.strategy.signal_strategy import WeightStrategyBase

root = Path.cwd().resolve()
while not (root / 'AGENTS.md').exists():
    if root.parent == root:
        raise RuntimeError('repo root not found')
    root = root.parent
if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.workflow import QlibNativeExperimentSpec, run_qlib_native_experiment

provider_uri = root / 'data' / 'qlib' / 'soxl'
qlib.init(provider_uri=provider_uri.as_posix(), region='us', kernels=1)
_calendar = D.calendar(start_time='2012-01-03', end_time='2026-06-09', freq='day')

card_dir = root / 'research' / 'qlib_signal_weight_search_v1'
spec = QlibNativeExperimentSpec(
    experiment='qlib_signal_weight_search_v1',
    provider_uri=provider_uri,
    recorder_uri=card_dir / 'qlib_recorder',
    output_dir=card_dir,
    market='all',
    region='us',
    instrument='SOXL',
    train=('2012-01-03', '2019-12-31'),
    test=('2020-01-02', '2026-06-09'),
)
result = run_qlib_native_experiment(spec)
summary = result.summary
legacy = summary['legacy_soxlbacktest_benchmark']
assert summary['feature_set']['source'] == 'Alpha158.get_feature_config+legacy_soxl_features'
assert summary['target_position_strategy']['class'] == 'SignalWeightStrategy'
# Search cards record the legacy comparison; promotion gates live in validated/production cards.

conclusion = f"""# 中文结论

- 做了什么：使用 qlib.init、D.calendar/D.features、DatasetH/DataHandlerLP、R.start、SignalRecord、SigAnaRecord、PortAnaRecord 和 WeightStrategyBase 子类 SignalWeightStrategy 跑 2012-2019 训练、2020-2026 测试的 SOXL 搜索卡。
- 模型逻辑：Alpha158 叠加旧 SOXL 特征，候选模型 {summary['model_selection']['candidate_count']} 个，选中 {summary['model_selection']['selected_model']} / {summary['model_selection']['selected_implementation']}。
- 核心指标：qlib 策略收益 {legacy['strategy_return_pct']:.2f}%，legacy soxl_backtest rulebase 收益 {legacy['legacy_rulebase_return_pct']:.2f}%，超额 {legacy['excess_return_pct']:.2f}%；feature_count={summary['feature_set']['feature_count']}，recorder_id={summary['recorder_id']}。
- regime/window 差异：regime_metrics.csv、window_metrics.csv、backtest_window_metrics.csv 和 risk_analysis.csv 为 qlib recorder 派生证据；测试窗口覆盖 2020-01-02 到 2026-06-09。
- 是否晋升：不直接晋升生产；本卡记录搜索诊断和 legacy 对比，validated/production 卡负责晋升门槛。
- 风险：单标的 timing 对阈值和样本切分敏感；当前搜索卡证明 qlib-native 路径和候选模型可超过 legacy，但不等于无未来函数的生产策略。
- 下一步：以 validated split 固定模型路径作为 production paper/sim 来源，并继续补 rolling/walk-forward 复验。
"""
(card_dir / 'conclusion.md').write_text(conclusion, encoding='utf-8')
print(json.dumps({
    'recorder_id': summary['recorder_id'],
    'selected_model': summary['model_selection'],
    'target_position_strategy': summary['target_position_strategy'],
    'legacy_soxlbacktest_benchmark': legacy,
}, indent=2, ensure_ascii=False))
```
