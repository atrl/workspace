# TSMC 月营收原始发布来源探针

只检查四个发行人新闻稿与两个 SEC 原始申报能否取得可哈希原文；不读价格、标签或训练。旧 SIA 三月平均全球销售额与库存/销售比路线不在本卡重开。

```python
from pathlib import Path
import json,hashlib
root=Path.cwd().resolve()
while not (root/"SPEC.md").is_file():
    if root.parent==root:raise RuntimeError("workspace root missing")
    root=root.parent
card=root/"research/tsmc_monthly_revenue_release_source_20261022"
plan=json.loads((card/"plan.json").read_text())
primary=json.loads((card/"retrieval_receipts.json").read_text())
secondary=json.loads((card/"sec_fallback_receipts.json").read_text())
browser=json.loads((card/"browser_fallback_receipt.json").read_text())
assert plan["frozen"] and len(primary["queries"])==4 and len(secondary["queries"])==2
assert all(x["http_status"]==403 for x in primary["queries"]+secondary["queries"])
for item in primary["queries"]+secondary["queries"]:
    path=root/item["relative_path"]
    assert hashlib.sha256(path.read_bytes()).hexdigest()==item["sha256"]
assert browser["status"]=="extension_not_connected" and browser["source_bytes_obtained"] is False
result={"status":"needs_source_access","decision":"no_TSM_monthly_factor_calculation","TSMC_official_press_responses":4,"SEC_6K_responses":2,"HTTP403_responses":6,"browser_extension_connected":False,"original_release_bytes_qualified":False,"first_public_time_qualified":False,"source_input_qualified":False,"Qlib_D_calls":0,"model_fits":0,"accounts":0,"market_outcome_reads":0,"SOXX_goal_achieved":False,"next_action":"Resume only with accessible exact original issuer or SEC bytes and filing/release timestamps; do not use search snippets or mutable current monthly table as historical PIT rows."}
(card/"source_probe_result.json").write_text(json.dumps(result,indent=2)+chr(10))
summary={"experiment":plan["experiment"],"status":result["status"],"decision":result["decision"],"result":"source_probe_result.json","recorder_status":"not_started_source_access_stop","objective_achieved":False,"qualified_forward_days":0,"model_fits":0,"accounts":0,"artifact_policy":plan["artifact_policy"]}
(card/"summary.json").write_text(json.dumps(summary,indent=2)+chr(10))
(card/"conclusion.md").write_text("# 中文结论\n\nTSMC 四个官方月营收原始新闻页和两个 SEC 6-K 原文链接在本机均返回 HTTP 403；浏览器桥接扩展未连接。六份响应原字节、URL、抓取时间和 SHA-256 已保留，但它们不是营收原文。网页搜索摘要不足以证明原始公告数值或首次公开时点。\n\n本路线标记 `needs_source_access`，没有读取 Qlib 价格/收益、拟合、账户或因子值。TSMC 月营收可能不同于已失败的 SIA 全球销售路径，但当前没有合格来源，不能称有效信号或 SOXX 超额。\n")
print(result["status"],result["HTTP403_responses"])
```
