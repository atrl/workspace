# 固定迁移实验归因结论

读取两张父卡的12个Qlib实际账户（各control/candidate×0/2/5bp），没有重新拟合、推理或回测。每账户944日；全部年度、季度以及819个连续126日窗口均保留。实际持仓四段收益、拆股/除息与原始费用重建日收益，最大误差5.03e-16；支付日仅转移应收款，不重复记收益。2bp为主分析，0/5bp为完整保留核对。

2bp结果：通用指数尾部迁移候选累计418.55%，配对对照392.04%。算术日收益加总差异中，平均暴露项−2.274个百分点，暴露与资产回报协变项+6.764个百分点，费用差异+0.033个百分点。家族混合训练候选累计238.26%，对照199.14%；平均暴露项−17.094个百分点、协变项+25.722个百分点、费用差异+0.814个百分点。前两项是在每种资产与持有时段内的population covariance恒等分解；这些数字不是累计收益百分点、不是恒定仓位反事实，也不证明可预测性。

收益差异明显依赖区间。指数迁移在2023部分年和2024分别落后配对对照约0.80和0.75个百分点，2025和2026部分年分别领先5.15和5.46个百分点；家族混合在2023部分年与2024落后约26.58和13.05个百分点，2025与2026部分年领先33.24和24.21个百分点。2023起点不完整，2026尚未结束；完整年按Qlib交易日覆盖核对，仅2024、2025。全部季度和126日窗口保留，不挑选胜出区间。先验趋势状态与内生账户风险状态分别汇总，不把事后胜负作为特征。

原始目标与实际开盘目标差异很大。例如指数候选原始SOXL目标平均比受控目标高17.98个百分点，原始SOXX低27.50个百分点；家族候选分别高13.63与低13.64个百分点。差异包含合法投影和账户风险路径，未运行控制器消融，不能归因于某一控制器组件。

复用原尾部评分Recorder：940个成熟标签、4个未知；候选logloss0.434972差于对照0.343716，Brier0.133975差于0.098389。候选预测均值27.47%，事件频率11.17%。因此“最终财富提高”不支持“风险预测更准确”；正协变项也不解决滚动稳定性、校准和独立验证问题。
