# 杠杆 ETF 在底层上涨时的稳定超额区间

比较 SOXL–SOXX 与 TQQQ–QQQ：按未来底层累计涨幅、持有期、路径效率和波动率，分析3x ETF相对1x ETF何时稳定产生超额。该条件使用未来实现收益，只是payoff map，不是入场信号。

```python
from pathlib import Path
import sys
import pandas as pd
import qlib
from qlib.config import REG_US
from qlib.data import D
from qlib.data.dataset import DatasetH
from qlib.workflow import R
from IPython.display import Markdown, display

repo_root = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(repo_root))
from src.workflow.leveraged_etf_positive_underlying_excess import run_leveraged_etf_positive_underlying_excess_card

source_audits = {}
for provider_name, symbols in [('soxl', ['SOXL', 'SOXX']), ('tqqq', ['TQQQ', 'QQQ'])]:
    provider_uri = repo_root / 'data' / 'qlib' / provider_name
    qlib.init(provider_uri=str(provider_uri), region=REG_US, kernels=1)
    calendar = D.calendar(freq='day')
    instruments = D.instruments('all')
    check = D.features(symbols, ['$close'], start_time='2010-03-11', end_time=str(pd.Timestamp(calendar[-1]).date()), freq='day')
    source_audits[provider_name] = {'calendar_rows': len(calendar), 'feature_rows': len(check), 'instruments': instruments}
source_audits
```
