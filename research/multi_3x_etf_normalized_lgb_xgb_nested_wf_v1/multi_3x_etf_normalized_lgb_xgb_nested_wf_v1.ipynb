## 中文结论

- 本轮不再选择人工模板；SOXL/TQQQ/UPRO/SPXL/TECL 先转成每个资产自己的 rolling z-score/percentile 特征，再 pooled 训练 LGB/XGB/consensus。
- 外层 SOXL 年份只用于最终报告，内层用 multi-asset leave-one-asset-out 选择候选；`selection_uses_outer_test=false`，`overfit_audit=mitigated`。
- 结果没有晋升：2023-2026 没有任何一年同时超过 SOXL buy-and-hold open-to-open 且回撤大于 -20%。失败主因不是回撤，而是模型把外层 SOXL 仓位压到约 6%-7%，收益显著低于 buy-and-hold。
- 这说明“补其他 3x ETF + 归一化 pooled LGB/XGB”是更正确的验证方向，但当前候选空间还没有达到目标，不能当作实盘模型。
- 下一步应预注册更多 3x ETF、扩大但固定候选仓位映射/风险预算，并继续 nested walk-forward；不能回到按年份挑模板。
- 不构成买卖建议。
