# ALFRED 两个历史版本日期的来源运输预检

```python
import json,hashlib
from pathlib import Path
card=Path.cwd()/'research/stage1_fed_h15_alfred_vintage_preflight_20261023'
plan=json.loads((card/'plan.json').read_text());result=json.loads((card/'transport_result.json').read_text())
sha=hashlib.sha256((card/'plan.json').read_bytes()).hexdigest()
assert plan['frozen'] and result['plan_sha256']==sha
assert result['selected_form']['vintage_dates']==plan['exact_request']['entered_vintage_dates']
assert result['download_attempts']==1 and not result['raw_vintage_ZIP_received']
assert not result['2011_vintage_proven'] and not result['2020_vintage_proven']
assert result['SOXX_market_outcome_reads']==result['Qlib_calls']==result['model_fits']==result['native_accounts']==0
print({'decision':result['decision'],'download_attempts':1,'raw_ZIP_received':False})
```
