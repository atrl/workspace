Hypothesis: SOXL/SOXX minute-derived 5min and 15min qlib providers may contain same-session next-bar volatility signal. This notebook runs the qlib-backed diagnostic card and does not promote the signal unless stability gates are proven.

```python
from pathlib import Path
import json
import sys
import pandas as pd
from IPython.display import display, Markdown

ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    for candidate in [Path.cwd().resolve(), *Path.cwd().resolve().parents]:
        if (candidate / "src").exists() and (candidate / "research").exists():
            ROOT = candidate
            break
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.workflow.intraday_bar_model import IntradayBarModelSpec, run_intraday_bar_model_card

EXPERIMENT = "agentic_alpha_intraday_5m15m_lgb_xgb_v1"
CARD = ROOT / "research" / EXPERIMENT
SPEC = IntradayBarModelSpec(
    experiment=EXPERIMENT,
    output_dir=CARD,
    recorder_uri=CARD / "qlib_recorder",
    provider_uris={
        "5min": ROOT / "data/qlib/soxl_intraday_5min",
        "15min": ROOT / "data/qlib/soxl_intraday_15min",
    },
)
result = run_intraday_bar_model_card(SPEC)
summary = result.summary
summary["recorder_id"], summary["selected_model"]
```
