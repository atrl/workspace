# 韩国半导体出口韩文公告来源交付

本卡是完整韩文来源档案的可恢复入口，原始字节在 NAS，数值化 Qlib 来源审查在关联 Recorder。这里不读取市场收益或拟合。

```python
from pathlib import Path
import json,hashlib
root=Path.cwd().resolve()
while not (root/"SPEC.md").is_file():
    if root.parent==root:raise RuntimeError("workspace root missing")
    root=root.parent
card=root/"research/korea_semiconductor_exports_korean_vintage_20261022"
plan=json.loads((card/"source_delivery_plan.json").read_text())
sha=lambda p:hashlib.sha256(Path(p).read_bytes()).hexdigest()
for path,digest in plan["identities_sha256"].items():assert sha(root/path)==digest,path
manifest=json.loads((card/"korean_national_article_manifest.json").read_text())
meta=json.loads((card/"korean_vintage_metadata_audit.json").read_text())
binding=json.loads((card/"nas_source_archive_binding.json").read_text())
full=json.loads((root/"research/korea_semiconductor_exports_korean_full_source_20261022/source_qualification_result.json").read_text())
assert manifest["months"]==meta["months"]==full["months"]==93
assert meta["PDF_versions"]==full["PDF_versions"]==96
assert binding["status"]=="archived_all_groups_remote_hash_verified" and len(binding["groups"])==9
assert binding["source_files"]==full["raw_original_files_archived"]==221
assert all(x["remote_SHA256_verified"] for x in binding["groups"].values())
result={"status":"Korean_93_month_original_source_archived_vintage_not_certified",
        "months":93,"PDF_versions":96,"double_version_months":sorted(meta["multiple_version_months"]),
        "source_files":221,"NAS_groups":9,"remote_hash_verified":True,
        "same_day_current_PDFs":meta["created_same_calendar_day"],
        "next_day_current_PDFs":meta["created_next_calendar_day"],
        "PDF_created_gt_7d_later":meta["created_gt_7_days_later"],
        "historical_PIT_qualified":False,"numeric_factor_qualified":False,
        "linked_native_Recorder_id":full["recorder_id"],
        "model_fits":0,"accounts":0,"market_outcome_reads":0,
        "next_action":"Do one separately frozen source-only numeric/duplicate-version audit; require contemporaneous original public vintages or prospective captures before a market information screen."}
(card/"source_delivery_result.json").write_text(json.dumps(result,indent=2)+chr(10))
summary={"experiment":"korea_semiconductor_exports_korean_vintage_20261022","status":result["status"],
         "decision":"source_coverage_archived_numeric_and_PIT_gates_pending",
         "result":"source_delivery_result.json","linked_native_Recorder_id":full["recorder_id"],
         "new_Recorder_status":"no_new_Recorder_reuses_full_source_card","model_fits":0,"accounts":0,
         "qualified_forward_days":0,"objective_achieved":False,
         "artifact_policy":{"policy":"slim_card_recorder_first","workflow_config":"workflow.yaml",
                            "recorder_only_artifacts":["sources/*","analysis/*"]}}
(card/"summary.json").write_text(json.dumps(summary,indent=2)+chr(10))
(card/"conclusion.md").write_text("# 中文结论\n\n韩文官方月报 2019-01 至 2026-09 的 93 篇原文和 96 个 PDF 版本已逐项验哈希，221 个来源文件分九组 NAS 归档并通过远端 SHA-256 核对；原生 Qlib 来源审查见 Recorder `"+full["recorder_id"]+"`。当前 PDF 中 50 个同日、13 个次日创建，12 个晚于网页登记一周以上，三个双版本月份全部保留。\n\n来源覆盖不等于历史首次公开版本合格；尚未提取可用的全月数值因子、未读取 SOXX 收益、没有拟合或账户。下一步只做单独冻结的数值与版本完整性审查，并继续累积未来原始发布收据。\n")
print(result["status"],result["months"],result["PDF_versions"])
```
