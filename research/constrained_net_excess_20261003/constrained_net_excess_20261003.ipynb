## 训练修复与开发期表现

## 后续审查发现的限制与未运行假设

当前市场特征门加全局theta预算，在固定市场状态下一般只能对单一持仓点精确HOLD，不能表示一片持仓区域内不交易；全局预算也限制不同状态分别走向全SOXL和全现金。详见[表达能力与既有实现审查](constrained_policy_expressivity_review.md)。这说明模型类存在限制，不证明扩大模型即可盈利。

另一个**拟合后、纯合成的导数审查**确认：旧计算图在零换手处可能产生错误的持仓状态导数，误差带有成本依赖（51个案例最大误差0.00248145），而费用、NAV和持仓的前向计算仍正确。修正原型利用费用后目标的精确归一化恒等式，恢复唯一状态导数；纯HOLD及非零换手部分控制本来就正确。标量费用/NAV在交易边界的单侧导数问题仍然存在。该修正**没有应用于本轮源码、检查点或结果**，也没有证明会提高收益；本轮训练轨迹不能被重标为已修正实现。见[状态导数审查](post_fee_state_gradient_audit.md)。

下一轮尚未运行的结构假设是保留原27→8市场编码器，增加四专家稀疏单纯形输出和直接持仓线性分支，参数251→272；通过精确嵌入保持原初始目标和持仓路径，同时允许学得HOLD区域。若同时采用已证明的状态导数修复，应在拟合前分别登记范围、源码和验证；保持原费用、风险、年度、种子及前向门。不能将这项未运行假设当成本轮成果。
