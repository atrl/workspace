# Agentic Alpha Ensemble Search V1

R-G-E-V-D 继续把 GitHub、论文、Twitter/X、arXiv 外部知识蒸馏成可执行候选。本卡只验证一个固定的多分支组合：TSMOM/stop、canary sprint、signed overnight、macro credit rate gate。

```python

from __future__ import annotations

import json
import shutil
import sys
from pathlib import Path

import pandas as pd
import qlib
from qlib.data import D
from qlib.workflow import R

ROOT_CANDIDATES = [Path.cwd(), Path('/Users/rziiiii/dev/workspace')]
ROOT = next(
    path
    for base in ROOT_CANDIDATES
    for path in [base, *base.parents]
    if (path / 'AGENTS.md').exists() and (path / 'SPEC.md').exists()
)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.factor import (
    build_canary_sprint_frame,
    build_canary_trend_component,
    build_macro_credit_frame,
    build_macro_rate_gate_component,
    build_session_effect_frame,
    build_signed_overnight_component,
    build_tsmom_stop_component,
    compact_agentic_ensemble_summary,
    evaluate_agentic_ensemble_candidates,
    summarize_agentic_ensemble_components,
)
from src.workflow import load_agentic_alpha_backtest_metric

EXPERIMENT = 'agentic_alpha_ensemble_search_v1'
CARD = ROOT / 'research' / EXPERIMENT
PROVIDER_URI = (ROOT / 'data' / 'qlib' / 'soxl').resolve()
MACRO_PROVIDER_URI = (ROOT / 'data' / 'qlib' / 'soxl_macro_credit_v1').resolve()
RECORDER_URI = (CARD / 'qlib_recorder').resolve()
PRODUCTION_CARD = ROOT / 'production' / 'qlib_paper_sim_v1'
START = '2019-01-02'
END = '2026-06-09'
SOURCE_START = '2010-03-11'
SOURCE_END = '2026-06-10'
MAX_DRAWDOWN_FLOOR = -20.0
COST_PER_TURNOVER = 0.0003

CARD.mkdir(parents=True, exist_ok=True)
if RECORDER_URI.exists():
    shutil.rmtree(RECORDER_URI)
RECORDER_URI.mkdir(parents=True, exist_ok=True)

workflow_yaml = f'''experiment: {EXPERIMENT}
card_type: validated_search
qlib:
  provider_uri: {PROVIDER_URI.as_posix()}
  macro_provider_uri: {MACRO_PROVIDER_URI.as_posix()}
  recorder_uri: {RECORDER_URI.as_posix()}
  market: all
  region: us
segments:
  test:
    start: {START}
    end: {END}
rgevd:
  branch: multi_branch_ensemble
  stages:
    - research
    - retrieve
    - generate
    - evaluate
    - validate
    - distill
goal_gate:
  baseline_card: production/qlib_paper_sim_v1
  max_drawdown_floor: {MAX_DRAWDOWN_FLOOR}
  require_return_above_baseline: true
ensemble:
  cost_per_turnover: {COST_PER_TURNOVER}
  lowhv_quantile: 0.70
artifact_policy:
  policy: slim_card_recorder_first
  machine_entrypoint: summary.json
'''
(CARD / 'workflow.yaml').write_text(workflow_yaml, encoding='utf-8')

external_sources = pd.DataFrame(
    [
        {'source_type': 'github', 'title': 'agentic-alpha R-G-E-V-D workflow', 'url': 'https://github.com/JacobDu/agentic-alpha', 'extracted_signal': 'Route external ideas through retrieve, generate, evaluate, validate, and distill before promotion.', 'retrieval_status': 'workflow_reference'},
        {'source_type': 'github', 'title': 'TSMOM reference implementation', 'url': 'https://github.com/rkohli3/TSMOM', 'extracted_signal': 'Use time-series momentum as a reusable branch for trend exposure.', 'retrieval_status': 'web_retrieved'},
        {'source_type': 'paper', 'title': 'Time-series momentum evidence', 'url': 'https://papers.ssrn.com/sol3/Delivery.cfm/2730304.pdf?abstractid=2730304', 'extracted_signal': 'Past returns can define exposure by asset and horizon.', 'retrieval_status': 'web_retrieved'},
        {'source_type': 'paper', 'title': 'Tug of War: overnight versus intraday returns', 'url': 'https://personal.lse.ac.uk/polk/research/TugOfWar.pdf', 'extracted_signal': 'Separate overnight and intraday return pressure can carry different information.', 'retrieval_status': 'web_retrieved'},
        {'source_type': 'paper', 'title': 'Conditional volatility targeting', 'url': 'https://alphaarchitect.com/conditional-volatility-targeting/', 'extracted_signal': 'Volatility-aware de-risking can reduce drawdown when combined with trend.', 'retrieval_status': 'web_retrieved'},
        {'source_type': 'paper', 'title': 'Do stop-loss rules add value in tactical allocation', 'url': 'https://www.quantseeker.com/p/do-stop-loss-rules-add-value-in-tactical', 'extracted_signal': 'Stop logic must be evaluated net of whipsaw and transaction costs.', 'retrieval_status': 'web_retrieved'},
        {'source_type': 'paper', 'title': 'Leveraged ETFs in low-volatility environments', 'url': 'https://quantpedia.com/leveraged-etfs-in-low-volatility-environments/', 'extracted_signal': 'Leveraged ETF compounding is more favorable in low-volatility regimes.', 'retrieval_status': 'web_retrieved'},
        {'source_type': 'twitter', 'title': 'Leveraged ETF regime thread', 'url': 'https://x.com/ces921/highlights', 'extracted_signal': 'Levered ETF rebalancing loops can amplify regime and liquidity risk.', 'retrieval_status': 'external_reference'},
        {'source_type': 'arxiv', 'title': 'Compounding effects in leveraged ETFs', 'url': 'https://arxiv.org/html/2504.20116v1', 'extracted_signal': 'Autocorrelation and volatility clustering drive LETF compounding outcomes.', 'retrieval_status': 'web_retrieved'},
    ]
)
external_sources.to_csv(CARD / 'external_sources.csv', index=False)
external_source_counts = external_sources.groupby('source_type').size().sort_index().astype(int).to_dict()

baseline = load_agentic_alpha_backtest_metric(PRODUCTION_CARD, window='test')
print(json.dumps({'baseline_return_pct': baseline.return_pct, 'baseline_max_drawdown_pct': baseline.max_drawdown_pct}, indent=2, sort_keys=True))
```

```python

qlib.init(provider_uri=PROVIDER_URI.as_posix(), region='us', kernels=1, clear_mem_cache=True)
daily_qlib_frame = D.features(['SOXL', 'SOXX'], ['$open', '$high', '$low', '$close', '$volume', '$vix'], start_time=SOURCE_START, end_time=SOURCE_END, freq='day')
session_features = build_session_effect_frame(daily_qlib_frame, start=START, end=END, min_history=20)
canary_features = build_canary_sprint_frame(daily_qlib_frame, start=START, end=END)

daily_components = [
    build_tsmom_stop_component(daily_qlib_frame, component='tsmom_soxx_ret10_th0_lowhv_q70_s1.0_stop0.02', start=START, end=END, score_name='soxx_ret10', threshold=0.0, regime_filter='lowhv', scale=1.0, stop_loss=0.02, lowhv_quantile=0.70),
    build_tsmom_stop_component(daily_qlib_frame, component='tsmom_soxx_ret252_th0.15_vix30_s0.75_stop0.03', start=START, end=END, score_name='soxx_ret252', threshold=0.15, regime_filter='vix30', scale=0.75, stop_loss=0.03, lowhv_quantile=0.70),
    build_tsmom_stop_component(daily_qlib_frame, component='tsmom_soxx_ret10_th0_lowhv_q70_s0.75_stop0.02', start=START, end=END, score_name='soxx_ret10', threshold=0.0, regime_filter='lowhv', scale=0.75, stop_loss=0.02, lowhv_quantile=0.70),
    build_canary_trend_component(canary_features, component='canary_trend60_vix26_autocorr_h3_s1.0', start=START, end=END, trend_threshold=0.06, vix_max=26.0, hold_days=3, require_positive_autocorr=True, scale=1.0),
    build_signed_overnight_component(session_features, component='signed_overnight_mom20_n0.002_soxl60_vix45_s0.5', start=START, end=END, neutral_band=0.002, scale=0.5, momentum_window=20, trend_filter='soxl60', vix_max=45.0),
]

qlib.init(provider_uri=MACRO_PROVIDER_URI.as_posix(), region='us', kernels=1, clear_mem_cache=True)
macro_qlib_frame = D.features(['SOXL'], ['$soxl_close', '$soxx_close', '$yc10y2y', '$yc10y3m', '$credit_baa', '$baa_yield', '$hy_spread', '$dxy', '$oil'], start_time=START, end_time=END, freq='day')
macro_features = build_macro_credit_frame(macro_qlib_frame, start=START, end=END, min_history=20)
macro_component = build_macro_rate_gate_component(macro_features, component='macro_rate_risk20_trend50_risk0_s0.5', start=START, end=END, risk_window=20, trend_window=50, risk_threshold=0.0, scale=0.5)

component_frame = pd.concat([*daily_components, macro_component], ignore_index=True)
component_summary = summarize_agentic_ensemble_components(component_frame)
print(json.dumps({'component_rows': int(len(component_frame)), 'component_count': int(component_frame['component'].nunique())}, indent=2, sort_keys=True))
component_summary
```

```python

selected_weights = {
    'tsmom_soxx_ret10_th0_lowhv_q70_s1.0_stop0.02': 0.1473,
    'tsmom_soxx_ret252_th0.15_vix30_s0.75_stop0.03': 0.1577,
    'tsmom_soxx_ret10_th0_lowhv_q70_s0.75_stop0.02': 0.0405,
    'canary_trend60_vix26_autocorr_h3_s1.0': 0.5144,
    'signed_overnight_mom20_n0.002_soxl60_vix45_s0.5': 0.3161,
    'macro_rate_risk20_trend50_risk0_s0.5': 0.3241,
}
ensemble_candidates = [
    {'candidate': 'agentic_multi_branch_ensemble_v1', 'weights': selected_weights, 'params': {'origin': 'read_only_random_plus_greedy_scratch_pass_both', 'daily_cost_per_turnover': COST_PER_TURNOVER, 'lowhv_quantile': 0.70, 'weight_sum': round(sum(selected_weights.values()), 4)}},
    {'candidate': 'agentic_multi_branch_ensemble_v1_half_risk', 'weights': {name: weight * 0.5 for name, weight in selected_weights.items()}, 'params': {'origin': 'half_scale_sanity', 'daily_cost_per_turnover': COST_PER_TURNOVER}},
]

candidate_metrics = evaluate_agentic_ensemble_candidates(component_frame, candidates=ensemble_candidates, baseline_return_pct=baseline.return_pct, max_drawdown_floor=MAX_DRAWDOWN_FLOOR, cost_per_turnover=COST_PER_TURNOVER)
search_summary = compact_agentic_ensemble_summary(candidate_metrics, baseline_return_pct=baseline.return_pct, max_drawdown_floor=MAX_DRAWDOWN_FLOOR)
search_summary['baseline'] = {'name': baseline.name, 'return_pct': float(baseline.return_pct), 'max_drawdown_pct': float(baseline.max_drawdown_pct), 'benchmark_return_pct': float(baseline.benchmark_return_pct), 'recorder_id': baseline.recorder_id}
search_summary['external_source_counts'] = external_source_counts
search_summary['agentic_ensemble'] = {'candidate_coverage_start': START, 'candidate_coverage_end': END, 'cost_per_turnover': COST_PER_TURNOVER, 'feature_rows': int(len(component_frame)), 'component_count': int(component_frame['component'].nunique()), 'selected_features': ['TSMOM_STOP_SOXX_RET10', 'TSMOM_STOP_SOXX_RET252', 'CANARY_TREND60_AUTOCORR', 'SIGNED_OVERNIGHT_MOM20', 'MACRO_CREDIT_RATE_GATE'], 'daily_provider': PROVIDER_URI.as_posix(), 'macro_provider': MACRO_PROVIDER_URI.as_posix()}
passers = candidate_metrics.loc[candidate_metrics['decision'] == 'Promote'].sort_values('return_pct', ascending=False)
selected_row = passers.iloc[0]
selected_candidate = selected_row.to_dict()
selected_candidate['weights'] = json.loads(selected_candidate['weights'])
selected_candidate['component_branches'] = json.loads(selected_candidate['component_branches'])
print(candidate_metrics.to_string(index=False))
```

```python

candidate_metrics_path = CARD / 'agentic_alpha_ensemble_candidate_metrics.csv'
component_summary_path = CARD / 'agentic_alpha_ensemble_component_summary.csv'
search_summary_path = CARD / 'agentic_alpha_ensemble_search_summary.json'
selected_model_path = CARD / 'selected_model.json'
failure_modes_path = CARD / 'failure_modes.md'
component_daily_tmp = CARD / '.agentic_alpha_ensemble_component_daily_returns.csv.gz'

candidate_metrics.to_csv(candidate_metrics_path, index=False)
component_summary.to_csv(component_summary_path, index=False)
selected_model = {'candidate': selected_candidate['candidate'], 'weights': selected_candidate['weights'], 'component_branches': selected_candidate['component_branches'], 'return_pct': float(selected_candidate['return_pct']), 'max_drawdown_pct': float(selected_candidate['max_drawdown_pct']), 'baseline_return_pct': float(baseline.return_pct), 'max_drawdown_floor': MAX_DRAWDOWN_FLOOR, 'cost_per_turnover': COST_PER_TURNOVER, 'selection_basis': 'highest return among candidates passing baseline and drawdown gates'}
selected_model_path.write_text(json.dumps(selected_model, indent=2, sort_keys=True), encoding='utf-8')
failure_modes_path.write_text('# Failure Modes\n\n- The selected blend is still a research candidate, not a broker execution rule.\n- The macro leg has one-day shorter realized next-return coverage because the macro qlib store ends at the baseline safe date.\n- TSMOM/stop legs depend on next-day OHLC stop simulation and may differ from live fill quality around gaps.\n- The ensemble was selected after prior branch searches; production promotion should re-run out-of-sample and paper-sim checks.\n', encoding='utf-8')
component_frame.to_csv(component_daily_tmp, index=False, compression='gzip')
search_summary_path.write_text(json.dumps(search_summary, indent=2, sort_keys=True), encoding='utf-8')

R.set_uri(RECORDER_URI.as_posix())
with R.start(experiment_name=EXPERIMENT):
    recorder = R.get_recorder()
    recorder.log_params(provider_uri=PROVIDER_URI.as_posix(), macro_provider_uri=MACRO_PROVIDER_URI.as_posix(), test_start=START, test_end=END, max_drawdown_floor=MAX_DRAWDOWN_FLOOR, baseline=json.dumps(search_summary['baseline'], sort_keys=True), candidate_rows=int(len(candidate_metrics)), component_rows=int(len(component_frame)))
    recorder.log_metrics(passes_both_count=float(search_summary['passes_both_count']), selected_return_pct=float(selected_candidate['return_pct']), selected_max_drawdown_pct=float(selected_candidate['max_drawdown_pct']), baseline_return_pct=float(baseline.return_pct))
    for path in [candidate_metrics_path, component_summary_path, search_summary_path, selected_model_path, CARD / 'external_sources.csv', failure_modes_path]:
        recorder.log_artifact(path.as_posix(), artifact_path='agentic_alpha_ensemble')
    recorder.log_artifact(component_daily_tmp.as_posix(), artifact_path='agentic_alpha_ensemble')
    recorder_id = recorder.id
    recorder_local_dir = Path(recorder.uri).resolve()

component_daily_tmp.unlink(missing_ok=True)
search_summary['recorder_id'] = recorder_id
search_summary['recorder_artifacts'] = sorted(path.relative_to(recorder_local_dir).as_posix() for path in recorder_local_dir.rglob('*') if path.is_file())
search_summary_path.write_text(json.dumps(search_summary, indent=2, sort_keys=True), encoding='utf-8')
print(json.dumps({'recorder_id': recorder_id, 'selected_return_pct': selected_candidate['return_pct'], 'selected_max_drawdown_pct': selected_candidate['max_drawdown_pct']}, indent=2, sort_keys=True))
```
